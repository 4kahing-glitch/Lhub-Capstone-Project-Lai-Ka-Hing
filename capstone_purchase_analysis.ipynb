```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load your datasets (using the raw GitHub URLs from your project)
url_customer = 'https://raw.githubusercontent.com/4kahing-glitch/Lhub-Capstone-Project-Lai-Ka-Hing/refs/heads/main/customer.csv'
url_purchase = 'https://raw.githubusercontent.com/4kahing-glitch/Lhub-Capstone-Project-Lai-Ka-Hing/refs/heads/main/purchase.csv'

df_customer = pd.read_csv(url_customer)
df_purchase = pd.read_csv(url_purchase)

# 2. Merge the datasets on 'customer_id'
df_merged = pd.merge(df_purchase, df_customer, on='customer_id', how='inner')

# 3. Calculate Total Revenue per item (Price * Quantity minus any discounts if applicable)
# Assuming total revenue = price * quantity
df_merged['total_revenue'] = df_merged['price'] * df_merged['quantity']

# --- VISUALIZATION 1: Total Revenue by Product Category ---
plt.figure(figsize=(10, 6))
category_revenue = df_merged.groupby('product_name')['total_revenue'].sum().sort_values(ascending=False).reset_index()

sns.barplot(data=category_revenue, x='total_revenue', y='product_name', palette='Blues_r')
plt.title('Total Revenue Contribution by Product Category', fontsize=14, fontweight='bold')
plt.xlabel('Total Revenue ($)', fontsize=12)
plt.ylabel('Product Category', fontsize=12)
plt.tight_layout()
plt.show()

# --- VISUALIZATION 2: Product Category Performance Across Gender Segments ---
plt.figure(figsize=(12, 6))
gender_category = df_merged.groupby(['product_name', 'gender'])['total_revenue'].sum().reset_index()

sns.barplot(data=gender_category, x='product_name', y='total_revenue', hue='gender', palette='Set2')
plt.title('Product Category Revenue Breakdown by Customer Gender', fontsize=14, fontweight='bold')
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Total Revenue ($)', fontsize=12)
plt.xticks(rotation=45)
plt.legend(title='Gender')
plt.tight_layout()
plt.show()

# --- VISUALIZATION 3: Sales Volume (Quantity Sold) by Country ---
plt.figure(figsize=(12, 6))
country_category = df_merged.groupby(['country', 'product_name'])['quantity'].sum().reset_index()

sns.barplot(data=country_category, x='country', y='quantity', hue='product_name', palette='tab10')
plt.title('Sales Volume (Quantity) of Product Categories Across Countries', fontsize=14, fontweight='bold')
plt.xlabel('Country', fontsize=12)
plt.ylabel('Total Quantity Sold', fontsize=12)
plt.xticks(rotation=45)
plt.legend(title='Product Category', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

###**Observation of Objective 1**

1. Total Revenue Contribution by Product Category
Observation: Revenue generation is fairly evenly distributed across the top product categories, though categories like Furniture and Rugs (driven by higher individual item prices) tend to pull in the highest cumulative revenue contributions.

2. Product Category Revenue Breakdown by Customer Gender
Observation: Spending habits and category preferences remain well-balanced between male and female customers, with both demographic segments showing comparable revenue shares across almost all product categories without any extreme gender-skewed buying spikes.

3. Sales Volume (Quantity) of Product Categories Across Countries
Observation: Sales volume (total units sold) is consistent across all operating countries (such as Brazil, Chile, and Mexico), indicating that international demand for your product categories scales uniformly rather than being dominated by a single region.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Assuming df_merged is already created from Objective 1,
# or we merge the datasets again to ensure it's self-contained:
url_customer = 'https://raw.githubusercontent.com/4kahing-glitch/Lhub-Capstone-Project-Lai-Ka-Hing/refs/heads/main/customer.csv'
url_purchase = 'https://raw.githubusercontent.com/4kahing-glitch/Lhub-Capstone-Project-Lai-Ka-Hing/refs/heads/main/purchase.csv'

df_customer = pd.read_csv(url_customer)
df_purchase = pd.read_csv(url_purchase)
df_merged = pd.merge(df_purchase, df_customer, on='customer_id', how='inner')

# Calculate basic financial metrics per order
df_merged['gross_revenue'] = df_merged['price'] * df_merged['quantity']
df_merged['discount_amount'] = df_merged['gross_revenue'] * df_merged['discount']
df_merged['net_revenue'] = df_merged['gross_revenue'] - df_merged['discount_amount']
df_merged['tax_amount'] = df_merged['net_revenue'] * df_merged['tax']

# --- VISUALIZATION 1: Impact of Unit Price on Order Quantity (Demand) ---
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_merged, x='price', y='quantity', alpha=0.5, color='teal')
plt.title('Impact of Unit Price on Order Quantities (Customer Demand)', fontsize=14, fontweight='bold')
plt.xlabel('Unit Price ($)', fontsize=12)
plt.ylabel('Order Quantity', fontsize=12)
plt.tight_layout()
plt.show()

# --- VISUALIZATION 2: Relationship Between Shipping Costs and Order Values ---
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_merged, x='shipping_cost', y='net_revenue', alpha=0.5, color='coral')
plt.title('Shipping Cost vs. Net Order Revenue', fontsize=14, fontweight='bold')
plt.xlabel('Shipping Cost ($)', fontsize=12)
plt.ylabel('Net Revenue ($)', fontsize=12)
plt.tight_layout()
plt.show()

# --- VISUALIZATION 3: Distribution of Taxes vs. Net Revenue by Product Category ---
plt.figure(figsize=(12, 6))
cat_tax = df_merged.groupby('product_name')[['net_revenue', 'tax_amount']].mean().reset_index()
cat_tax_melted = cat_tax.melt(id_vars='product_name', value_vars=['net_revenue', 'tax_amount'],
                              var_name='Metric', value_name='Average Amount ($)')

sns.barplot(data=cat_tax_melted, x='product_name', y='Average Amount ($)', hue='Metric', palette='Set1')
plt.title('Average Net Revenue vs. Tax Amount per Product Category', fontsize=14, fontweight='bold')
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Average Amount ($)', fontsize=12)
plt.xticks(rotation=45)
plt.legend(title='Financial Metric')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the dataset
url_purchase = 'https://raw.githubusercontent.com/4kahing-glitch/Lhub-Capstone-Project-Lai-Ka-Hing/refs/heads/main/purchase.csv'
df_purchase = pd.read_csv(url_purchase)

# 2. Convert date columns to datetime format & extract temporal features
df_purchase['order_date'] = pd.to_datetime(df_purchase['order_date'], format='%d-%m-%Y')
df_purchase['shipping_date'] = pd.to_datetime(df_purchase['shipping_date'], format='%d-%m-%Y')

df_purchase['year'] = df_purchase['order_date'].dt.year
df_purchase['month'] = df_purchase['order_date'].dt.month
df_purchase['year_month'] = df_purchase['order_date'].dt.to_period('M')

# Calculate fulfillment time (in days) between order date and shipping date
df_purchase['fulfillment_days'] = (df_purchase['shipping_date'] - df_purchase['order_date']).dt.days

# Calculate net revenue per order for trend analysis
df_purchase['gross_revenue'] = df_purchase['price'] * df_purchase['quantity']
df_purchase['net_revenue'] = df_purchase['gross_revenue'] * (1 - df_purchase['discount'])


# --- VISUALIZATION 1: Seasonal Purchasing Peaks (Monthly Revenue Trend) ---
plt.figure(figsize=(12, 6))
monthly_trend = df_purchase.groupby('month')['net_revenue'].sum().reset_index()

sns.lineplot(data=monthly_trend, x='month', y='net_revenue', marker='o', color='b', linewidth=2.5)
plt.title('Seasonal Purchasing Peaks: Total Net Revenue by Month', fontsize=14, fontweight='bold')
plt.xlabel('Month of the Year', fontsize=12)
plt.ylabel('Total Net Revenue ($)', fontsize=12)
plt.xticks(ticks=range(1, 13), labels=['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()


# --- VISUALIZATION 2: Long-Term Sales Trends Over Time (Year-Month) ---
plt.figure(figsize=(14, 6))
timeline_trend = df_purchase.groupby('year_month')['net_revenue'].sum().reset_index()
timeline_trend['year_month_str'] = timeline_trend['year_month'].astype(str)

sns.barplot(data=timeline_trend, x='year_month_str', y='net_revenue', palette='viridis')
plt.title('Long-Term Sales Trends (Monthly Net Revenue Over Time)', fontsize=14, fontweight='bold')
plt.xlabel('Year-Month', fontsize=12)
plt.ylabel('Net Revenue ($)', fontsize=12)
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()


# --- VISUALIZATION 3: Fulfillment Efficiency Over Time ---
plt.figure(figsize=(12, 6))
fulfillment_trend = df_purchase.groupby('year_month')['fulfillment_days'].mean().reset_index()
fulfillment_trend['year_month_str'] = fulfillment_trend['year_month'].astype(str)

sns.lineplot(data=fulfillment_trend, x='year_month_str', y='fulfillment_days', marker='s', color='darkorange', linewidth=2)
plt.title('Fulfillment Efficiency: Average Days to Ship Orders Over Time', fontsize=14, fontweight='bold')
plt.xlabel('Year-Month', fontsize=12)
plt.ylabel('Avg Fulfillment Duration (Days)', fontsize=12)
plt.xticks(rotation=90)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

1. Seasonal Purchasing Peaks (Monthly Revenue Trend)
Observation: Sales are spread out fairly evenly throughout the year without massive spikes during specific months. This shows steady, consistent customer purchasing habits month-over-month.

2. Long-Term Sales Trends Over Time (Year-Month)
Observation: Monthly total revenue remains stable across the entire multi-year timeline, indicating consistent business performance and steady demand without sharp market growth or sudden drops.

3. Fulfillment Efficiency (Average Days to Ship Orders Over Time)
Observation: The time it takes from placing an order to shipping remains consistent and stable over time, showing reliable warehouse and fulfillment performance.



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the dataset
url_purchase = 'https://raw.githubusercontent.com/4kahing-glitch/Lhub-Capstone-Project-Lai-Ka-Hing/refs/heads/main/purchase.csv'
df_purchase = pd.read_csv(url_purchase)

# 2. Calculate net revenue and prepare inventory metrics
df_purchase['gross_revenue'] = df_purchase['price'] * df_purchase['quantity']
df_purchase['net_revenue'] = df_purchase['gross_revenue'] * (1 - df_purchase['discount'])


# --- VISUALIZATION 1: Order Frequency by Product (Demand Popularity) ---
plt.figure(figsize=(12, 6))
product_freq = df_purchase['product_name'].value_counts().reset_index()
product_freq.columns = ['product_name', 'order_count']

sns.barplot(data=product_freq, x='order_count', y='product_name', palette='crest')
plt.title('Inventory Demand: Order Frequency by Product', fontsize=14, fontweight='bold')
plt.xlabel('Total Number of Orders', fontsize=12)
plt.ylabel('Product Name', fontsize=12)
plt.tight_layout()
plt.show()


# --- VISUALIZATION 2: Average Order Quantity per Product (Batch Size) ---
plt.figure(figsize=(12, 6))
avg_qty = df_purchase.groupby('product_name')['quantity'].mean().sort_values(ascending=False).reset_index()

sns.barplot(data=avg_qty, x='quantity', y='product_name', palette='magma')
plt.title('Stock Alignment: Average Order Quantity per Product', fontsize=14, fontweight='bold')
plt.xlabel('Average Units Bought per Order', fontsize=12)
plt.ylabel('Product Name', fontsize=12)
plt.tight_layout()
plt.show()


# --- VISUALIZATION 3: Total Sales Volume vs. Revenue Contribution (ABC Inventory Analysis Proxy) ---
plt.figure(figsize=(12, 6))
abc_analysis = df_purchase.groupby('product_name').agg(
    total_quantity=('quantity', 'sum'),
    total_revenue=('net_revenue', 'sum')
).reset_index()

sns.scatterplot(data=abc_analysis, x='total_quantity', y='total_revenue', hue='product_name', s=150, palette='tab10')
plt.title('Inventory Portfolio: Total Quantity Sold vs. Net Revenue Contribution', fontsize=14, fontweight='bold')
plt.xlabel('Total Units Sold (Volume)', fontsize=12)
plt.ylabel('Total Net Revenue ($)', fontsize=12)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title='Product')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

1.   Order Frequency by Product: Order transaction counts are well-distributed across product types, indicating that customers regularly purchase across the full product catalog rather than focusing on a single item.

2. Average Order Quantity per Product: The average number of units purchased per transaction remains consistent across categories, helping you plan baseline stock buffer levels without unexpected bulk demands.

3. Quantity Sold vs. Revenue Contribution: High-volume items align logically with overall revenue, helping you identify core products that require prioritized inventory turnover monitoring to reduce carrying costs.




```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the datasets
url_customer = 'https://raw.githubusercontent.com/4kahing-glitch/Lhub-Capstone-Project-Lai-Ka-Hing/refs/heads/main/customer.csv'
url_purchase = 'https://raw.githubusercontent.com/4kahing-glitch/Lhub-Capstone-Project-Lai-Ka-Hing/refs/heads/main/purchase.csv'

df_customer = pd.read_csv(url_customer)
df_purchase = pd.read_csv(url_purchase)

# 2. Merge datasets and calculate financial metrics
df_merged = pd.merge(df_purchase, df_customer, on='customer_id', how='inner')

df_merged['gross_revenue'] = df_merged['price'] * df_merged['quantity']
df_merged['discount_amount'] = df_merged['gross_revenue'] * df_merged['discount']
df_merged['net_revenue'] = df_merged['gross_revenue'] - df_merged['discount_amount']
# Profitability proxy: Net Revenue minus Shipping Costs
df_merged['net_profit'] = df_merged['net_revenue'] - df_merged['shipping_cost']


# --- VISUALIZATION 1: High-Value Customer Segmentation (Top Customers by Net Profit) ---
plt.figure(figsize=(12, 6))
top_customers = df_merged.groupby(['customer_id', 'first_name', 'last_name', 'country'])['net_profit'].sum().reset_index()
top_customers['full_name'] = top_customers['first_name'] + ' ' + top_customers['last_name']
top_customers_sorted = top_customers.sort_values(by='net_profit', ascending=False).head(10)

sns.barplot(data=top_customers_sorted, x='net_profit', y='full_name', hue='country', palette='coolwarm')
plt.title('Profitability Strategy: Top 10 High-Value Customers by Net Profit', fontsize=14, fontweight='bold')
plt.xlabel('Total Net Profit ($)', fontsize=12)
plt.ylabel('Customer Name', fontsize=12)
plt.legend(title='Country')
plt.tight_layout()
plt.show()


# --- VISUALIZATION 2: Product Pricing & Profit Margin Analysis ---
plt.figure(figsize=(12, 6))
pricing_profit = df_merged.groupby('product_name').agg(
    avg_price=('price', 'mean'),
    avg_discount=('discount', 'mean'),
    avg_net_profit=('net_profit', 'mean')
).sort_values(by='avg_net_profit', ascending=False).reset_index()

sns.barplot(data=pricing_profit, x='avg_net_profit', y='product_name', palette='mako')
plt.title('Pricing Strategy: Average Net Profit per Product Category', fontsize=14, fontweight='bold')
plt.xlabel('Average Net Profit ($)', fontsize=12)
plt.ylabel('Product Category', fontsize=12)
plt.tight_layout()
plt.show()


# --- VISUALIZATION 3: Logistics Optimization (Shipping Cost vs. Net Revenue Ratio) ---
plt.figure(figsize=(10, 6))
df_merged['shipping_ratio'] = (df_merged['shipping_cost'] / df_merged['net_revenue']) * 100
shipping_opt = df_merged.groupby('product_name')['shipping_ratio'].mean().sort_values(ascending=False).reset_index()

sns.barplot(data=shipping_opt, x='shipping_ratio', y='product_name', palette='rocket')
plt.title('Logistics Efficiency: Average Shipping Cost as a % of Net Revenue', fontsize=14, fontweight='bold')
plt.xlabel('Shipping Cost Ratio (%)', fontsize=12)
plt.ylabel('Product Category', fontsize=12)
plt.tight_layout()
plt.show()
```

**Observations for Objective 5**
1. High-Value Customer Groups: Identifying top-tier profitable customers allows the business to design loyalty programs and targeted marketing campaigns tailored to high-value buyer segments.

2. Product Pricing Optimization: Comparing net profit margins across categories reveals which products drive the strongest bottom-line returns, helping guide discount strategies and price adjustments.

3. Logistics & Fulfillment Efficiency: Measuring shipping costs relative to net revenue exposes fulfillment overhead margins, allowing you to optimize delivery routes or adjust shipping fee thresholds to protect profitability.
