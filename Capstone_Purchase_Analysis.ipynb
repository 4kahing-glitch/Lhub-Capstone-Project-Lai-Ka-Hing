```python
import pandas as pd

url1 = 'https://raw.githubusercontent.com/4kahing-glitch/Lhub-Capstone-Project-Lai-Ka-Hing/refs/heads/main/customer.csv'
url2 = 'https://raw.githubusercontent.com/4kahing-glitch/Lhub-Capstone-Project-Lai-Ka-Hing/refs/heads/main/purchase.csv'
df = pd.read_csv(url1)
print(df.head())

import numpy as np

# Check for missing values
print("Missing values per column initially:")
print(df.isnull().sum())


import matplotlib.pyplot as plt
import seaborn as sns

# Select only numeric columns and calculate the correlation matrix
numeric_cols = df.select_dtypes(include=[np.number]).columns
corr_matrix = df[numeric_cols].corr()

print("Correlation Matrix:")
display(corr_matrix)

# Plot correlation heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix of Numeric Columns with Respect to Sales')
plt.show()
```
